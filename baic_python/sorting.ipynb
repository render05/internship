```python
data = {
    'one':["a","b","c","d","a","b",],
    'two':[2,42,42,6,7,8]
}
df = pd.DataFrame(data)
# res = df.groupby('one')['two'].sum()
# res = df.groupby('one')['two'].mean()
# res = df.groupby('one')['two'].min()
res = df.groupby('one')['two'].agg(['sum','mean'])

print(res)
```
