```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label='sin(x)', color='red', linestyle='--')

plt.title("Sin Values")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.legend()
plt.grid()

plt.show()
```
