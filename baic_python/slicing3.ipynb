```python
import pandas as pd 
df = pd.DataFrame({
    "col1": range(12),
    "col2": ['A']*3 + ['B']*3+['C']*3 + ['D']*3,
    "date": pd.to_datetime(['2026-05-27','2026-05-26','2026-05-22']*4)
    })
print(df)
pivoted =df.pivot(index='date',columns='col2',values= 'col1')
print(pivoted)
```
