## 2) Basic Math Operations (using a function)

```python
def basic_math_operations(a, b):
    """Return basic math results for two numbers.

    Operations: add, subtract, multiply, divide
    """
    results = {}
    results['add'] = a + b
    results['subtract'] = a - b
    results['multiply'] = a * b

    if b == 0:
        results['divide'] = 'Division by zero is not allowed'
    else:
        results['divide'] = a / b

    return results

# Take 2 number inputs from the user
a = float(input('Enter first number (a): '))
b = float(input('Enter second number (b): '))

outputs = basic_math_operations(a, b)
print('
Results:')
for op, val in outputs.items():
    print(f'{op}: {val}')
```

## 3) Check Palindrome Number

```python
def is_palindrome_number(n):
    """Check whether a number is a palindrome.

    Steps:
    1) Find reverse of the given number
    2) Compare that number with the reverse number
    3) If equal => Palindrome else not
    """
    n_int = int(n)

    # Palindrome numbers are usually non-negative; handle negative safely
    if n_int < 0:
        return False

    original = n_int
    reverse = 0

    while original > 0:
        digit = original % 10
        reverse = reverse * 10 + digit
        original //= 10

    return n_int == reverse

num = int(input('Enter a number to check palindrome: '))

if is_palindrome_number(num):
    print(num, 'is a Palindrome Number')
else:
    print(num, 'is NOT a Palindrome Number')
```
