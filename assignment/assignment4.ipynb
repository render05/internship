```python
# 1) Maximum of three numbersdef max_of_three(a, b, c):    """Return the maximum of three numbers."""    return max(a, b, c)
```

```python
# 2) Distinct elements from a list (preserve order)def distinct_elements(lst):    """Return a new list with distinct elements from the first list."""    seen = set()    result = []    for x in lst:        if x not in seen:            seen.add(x)            result.append(x)    return result
```

```python
# 3) Multiply all numbers in a listdef multiply_all(lst):    """Multiply all the numbers in a list."""    product = 1    for x in lst:        product *= x    return product
```

```python
# 4) Factorial of a non-negative integerdef factorial(n):    """Calculate factorial of a non-negative integer n."""    if n < 0:        raise ValueError('factorial() is defined only for non-negative integers')    result = 1    for i in range(2, n + 1):        result *= i    return result
```

```python
# 5) Reverse a stringdef reverse_string(s):    """Reverse a string."""    return s[::-1]
```

```python
# 6) Check whether a number falls within a given range (inclusive)def in_range(number, low, high):    """Return True if number is within [low, high] inclusive, else False."""    return low <= number <= high
```

```python
# 7) Print Even Numbers from a Given Listdef even_numbers(lst):    """Return a list containing only even numbers from the given list."""    return [x for x in lst if x % 2 == 0]
```

```python
# 8) Check whether a number is primedef is_prime(n):    """Return True if n is prime, otherwise False."""    if n < 2:        return False    if n == 2:        return True    if n % 2 == 0:        return False    i = 3    while i * i <= n:        if n % i == 0:            return False        i += 2    return True
```

```python
# 9) Count uppercase and lowercase letters in a stringdef count_case_letters(s):    """Accept a string and count uppercase and lowercase letters.
    Returns (upper_count, lower_count)."""    upper_count = 0    lower_count = 0    for ch in s:        if 'A' <= ch <= 'Z':            upper_count += 1        elif 'a' <= ch <= 'z':            lower_count += 1    return upper_count, lower_count
```

## File practice: writing and appending data

```python
def write_practice_file():    """Practise writing/appending data in a file."""    path = 'assignment4_data.txt'    # Write mode (creates/overwrites)    with open(path, 'w', encoding='utf-8') as f:        f.write('Assignment 4 - File Practice\n')        f.write("Initial content written using 'w' mode.\n")    # Append mode    with open(path, 'a', encoding='utf-8') as f:        f.write("Appended content using 'a' mode.\n")    return path
```
