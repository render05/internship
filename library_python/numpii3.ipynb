```python
#concatenate joining arrays 
import numpy as np
a = np.array([[1,2,3],[6,8,9]])
b = np.array([[7,4,9],[2,5,6]])



# res = np.concatenate((a,b))
res = np.concatenate((a,b),axis =1)
print(res)# double bracket dena taki baki parameters bhi use kar sakte hain ex axis 
```

```python
#splitting arrays 
a = np.arange(9)
print(a)#0 se 9 
res  = np.split(a,3)
print(res)# 3,3 ke array main divide kar dega 
for i , arr in  enumerate(res):
    print("spliited array",i+1)# splitted array sundar likhne ka tarika
    print(arr)#
```
